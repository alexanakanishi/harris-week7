```python
import pandas as pd

df = pd.read_csv("data/ethan_final_key_df_head.csv")
roads = pd.read_csv("roads1.csv")
ice = pd.read_csv("data/ice_covariates.csv")
census = pd.read_csv("data/mexico_census_infrastructure_data.csv")
schools = pd.read_csv("data/school_infrastructure_covariates.csv")
```

```python
df.head()
```

```python
roads.head()
```

```python
roads = roads.drop(columns=['municipio'])
```

```python
ice.head()
```

```python
ice = ice.drop(columns = ['NOMGEO'])
```

```python
census.head()
```

```python
census = census.drop(columns = ['NOMGEO'])
```

```python
schools.head()
```

```python
schools = schools.drop(columns = ['NOMGEO'])
```

```python
print(df["CVEGEO"].nunique())
print(roads["CVEGEO"].nunique())
print(ice["CVEGEO"].nunique())
print(census["CVEGEO"].nunique())
print(schools["CVEGEO"].nunique())
```

```python
anti_join1 = df[~df["CVEGEO"].isin(roads["CVEGEO"])]

print(anti_join1["CVEGEO"].nunique())

anti_join2 = roads[~roads["CVEGEO"].isin(df["CVEGEO"])]

print(anti_join2["CVEGEO"].nunique())
print(anti_join2["CVEGEO"].unique())

print(anti_join2)
```

```python
codes_to_drop = [9998, 14998, 20999, 22998, 26998, 26999, 30998, 30999]
roads = roads[~roads["CVEGEO"].isin(codes_to_drop)]
```

```python
print(df["CVEGEO"].nunique())
print(roads["CVEGEO"].nunique())
print(ice["CVEGEO"].nunique())
print(census["CVEGEO"].nunique())
print(schools["CVEGEO"].nunique())
```

```python
merged = df.merge(roads, on="CVEGEO", how="outer") \
           .merge(ice, on="CVEGEO", how="outer") \
           .merge(census, on="CVEGEO", how="outer") \
           .merge(schools, on="CVEGEO", how="outer")
```
