```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

data = pd.read_csv("data/Datos_viales_2013_2024.csv")
municipios = gpd.read_file("data/Mexico_mun_layer_simplified.shp")  # INEGI shapefile

# Create GeoDataFrame from your data
gdf = gpd.GeoDataFrame(
    data,
    geometry=gpd.points_from_xy(data.longitud, data.latitud),
    crs="EPSG:4326"
)

# Reproject to match shapefile CRS if needed
municipios = municipios.to_crs("EPSG:4326")

# Spatial join
result = gpd.sjoin(gdf, municipios[["CVEGEO", "NOMGEO", "geometry"]], how="left", predicate="within")
result = result.rename(columns={"NOMGEO": "municipio"})
result.drop(columns=["geometry", "index_right"])
```

Claude notation

The highway CSV has no municipality column — it only knows which road segment it is and what state it's in. The core idea is to extract place names hidden in the text fields and look them up in a municipality directory.

The approach has three stages:

1. Mining place names from text
Highway segment names like "Monterrey-Saltillo" or "Acambaro - Celaya" are essentially origin-destination pairs. The place names at either end of a segment are usually towns or cities, which are often (though not always) municipality names. The STRIP regex cleans out highway jargon (T. Izq., Lim. Edos., etc.) that looks like text but isn't a place, leaving just the meaningful geographic tokens.

2. Fuzzy matching against a known list
Rather than looking up exact strings (which would fail on accents, abbreviations, and spelling variants), you compare each extracted token against every municipality name in the INEGI catalog using a similarity score. The token_sort_ratio scorer is forgiving about word order and minor spelling differences. You get back the best-matching municipality and a confidence score, and only accept matches above a threshold.

3. Resolving ambiguity with a hierarchy
The same city name can appear in multiple states, so the code searches within the segment's state first (more precise), and only falls back to a nationwide search if nothing matches locally. Manual corrections handle the cases where fuzzy matching fundamentally can't work — abbreviations like "Cd. Valles" or localities like "Zihuatanejo" whose municipality has a different official name.

The fundamental limitation is that a road segment is a line, not a point — it can cross multiple municipalities, and you're assigning it to just one based on whichever place name matched best. You're also only recovering municipality identity from the endpoints named in the segment label, not from the segment's actual geographic path. This means the resulting CVE_MUN is an approximation: good enough for aggregating infrastructure covariates to the municipality level, but not a precise spatial attribution.

Honestly, the match quality is mixed and varies systematically by case type. Here's a realistic breakdown:

High confidence (~correct most of the time)

Segments where tramo or carretera starts with a major city name that is also the municipality name — "Acambaro - Celaya", "Monterrey - Saltillo". These are likely correct.
Manual corrections (corrections dict) — those were hand-verified, so score 100 means it's right.
High-scoring within-state matches (score ≥ 90) where the place name is unambiguous in that state.

Moderate confidence (~right but not guaranteed)

Scores in the 75–89 range — fuzzy matching at this level catches accent/spelling variants but also produces false positives, especially for short or common-sounding names.
Nationwide fallback matches — correct when the city is unambiguous nationally (e.g. "Monterrey"), but risky for common names that appear in multiple states (e.g. "San Juan", "Santa María", "Guadalupe").

Low confidence / likely wrong

Segment crosses multiple municipalities — you assigned one endpoint's municipality, but a 50km segment may pass through 4 or 5. This is structural, not a code bug.
City ≠ municipality name — "Zihuatanejo" is in municipality "José Azueta". You caught some with the corrections dict but there are others you didn't (e.g. "Cancún" → "Benito Juárez", "Los Cabos", "Valle de México").
Threshold 70 false positives — at this threshold, short names like "Rayon", "Uman", "Conkal" could match the wrong municipality in a different state during the nationwide fallback.
Infrastructure labels that survived STRIP — "Periférico de Mérida" matching to "Mérida" is probably fine, but "Libramiento" or "Ramal" matching anything is wrong.

Bottom line for your purposes

For a smoothness battery in an RDD, what matters is not that every segment is perfectly geocoded, but that misassignment is random with respect to the running variable (your reconstructed selection index). If mismatches are concentrated in certain highway types or regions that correlate with the selection score, they could contaminate the balance test. Otherwise, they just add noise and attenuate your covariate balance statistics toward zero — which is conservative, not invalidating.

The safest approach is to only use municipality-level aggregates where you have sufficient segment coverage (e.g., at least 3–5 matched segments per municipality), flag municipalities with low coverage in your technical appendix, and note that the attribution is approximate.

```python
from rapidfuzz import process, fuzz

# highway data
df = pd.read_csv("data/capacidad_niveles_servicio_red_carretera_federal.csv")

# INEGI municipality catalog
mun_cat = pd.read_csv("data/catun_municipio/AGEEML_20268211656753_utf8.csv", dtype=str)

# find state names in df that don't match INEGI official names
print(set(df["estado"]) - set(mun_cat["NOM_ENT"].unique()))
```

```python
# state name fixes
estado_fixes = {
    "Veracruz":        "Veracruz de Ignacio de la Llave",
    "Yucatan":         "Yucatán",
    "Ciudad de Mexico": "Ciudad de México",
    "Coahuila":        "Coahuila de Zaragoza",
    "Mexico":          "México",
    "San Luis Potosi": "San Luis Potosí",
    "Nuevo Leon":      "Nuevo León",
    "Michoacan":       "Michoacán de Ocampo",
    "Queretaro":       "Querétaro",
}

# estado_ent new column with corrected names
df["estado_ent"] = df["estado"].replace(estado_fixes)

print(set(df["estado_ent"]) - set(mun_cat["NOM_ENT"].unique()))
```

```python
import re
import unicodedata

# strip accents
def strip_accents(s):
    return ''.join(c for c in unicodedata.normalize('NFD', s)
                   if unicodedata.category(c) != 'Mn')

# Expanded strip pattern — now also catches T.C., X., and Lim. Edos fully
STRIP = r"""(?ix)
    T\.\s*C\.\s*                    |   # T. C. (troncal conexión)
    T\.\s*[IiDd][zqr][q\.]?\s*     |   # T. Izq., T. Der.
    X\.\s*C\.\s*                    |   # X. C.
    X\.\s+                          |   # X. (standalone)
    Lim\.\s*Edos?\.?[^-]*          |   # Lim. Edos. Term. ... (whole phrase)
    Plaza\s+de\s+Cobro[^-]*        |
    Libramiento\s+[^-]*de\s+       |
    Entronque[^-]*                  |
    Ent\.[^-]*                      |
    Acceso[^-]*                     |
    Ppia\.[^-]*                     |
    Term\.[^-]*
"""

def extract_places(s):
    cleaned = re.sub(STRIP, "", s)  # remove jargon (like TC etc)
    parts = [p.strip() for p in re.split(r"[-–]", cleaned) if p.strip()]
    return [p for p in parts if len(p) > 3 and not re.match(r"^\d", p)]


# apply extract places to both tramo (stretch) and carretera (road)
df["_places"]          = df["tramo"].apply(extract_places)
df["_carretera_places"] = df["carretera"].apply(extract_places)  # now uses STRIP too
df["_all_places"]       = df["_places"] + df["_carretera_places"]




# Normalize mun catalog — strip accents for matching, keep originals for output
mun_cat["_NOM_MUN_norm"] = mun_cat["NOM_MUN"].apply(lambda s: strip_accents(s).upper())

# Rebuild state_muns using accent-stripped keys
# building dictionary
state_muns = (
    mun_cat
    .groupby("NOM_ENT")
    .apply(lambda g: dict(zip(g["_NOM_MUN_norm"], g["CVEGEO"])))
    .to_dict()
)


# states missing after estado_fixes are excluded here
estado_map = {row: row for row in df["estado_ent"].unique()
              if row in mun_cat["NOM_ENT"].values}



# Nationwide lookup (all states collapsed)
# used when within-state search fails
all_muns = dict(zip(mun_cat["_NOM_MUN_norm"], mun_cat["CVEGEO"]))

corrections = {
    "CD. VALLES":              "24035",  # Ciudad Valles, SLP
    "TUXPAM":                  "30185",  # Tuxpan, Veracruz
    "ZIHUATANEJO":             "12038",  # José Azueta, Guerrero
    "PUERTO JUAREZ":           "23005",  # Benito Juárez (Cancún), Q. Roo
    "PERIFERICO DE MERIDA":    "31050",  # Mérida, Yucatán
    "PERIFERICO DE SAN LUIS POTOSI": "24028",  # SLP capital
    "VENTA DE CARPIO":         "15033",  # Ecatepec, México
    "TIHUATLAN":               "30175",  # Tihuatlán, Veracruz
    "CD. JUAREZ":              "08037",  # Juárez, Chihuahua
}



def match_place_to_mun(place, estado, threshold=70):
    ent = estado_map.get(estado)
    query = strip_accents(place).upper()
    
    if query in corrections:
        return corrections[query], 100
    
    # First try within-state (more precise)
    if ent:
        candidates = state_muns.get(ent, {})
        result = process.extractOne(query, list(candidates.keys()),
                                    scorer=fuzz.token_sort_ratio)
        if result and result[1] >= threshold:
            return candidates[result[0]], result[1]
    
    # Fall back to nationwide search
    result = process.extractOne(query, list(all_muns.keys()),
                                scorer=fuzz.token_sort_ratio)
    if result and result[1] >= threshold:
        return all_muns[result[0]], result[1]
    return None, None
```

```python
# Resolve each segment to a CVEGEO 
# for each road segment, tries every extracted place name and keeps whiever one produces the highest confidence municipality match. one CVE_MUN per segment
records = []
for _, row in df.iterrows():
    best_cvegeo, best_score = None, 0
    for place in row["_all_places"]:
        cvegeo, score = match_place_to_mun(place, row["estado"])
        if score and score > best_score:
            best_cvegeo, best_score = cvegeo, score
    records.append({"consecutivo": row["consecutivo"],
                    "CVE_MUN": best_cvegeo,
                    "match_score": best_score})

matches = pd.DataFrame(records)

df = df.merge(matches, on="consecutivo", how="left")

# Audit: how many got assigned?
print(df["CVE_MUN"].notna().value_counts())
# Review low-confidence matches
df[df["match_score"].between(75, 85)][["tramo", "estado", "CVE_MUN", "match_score"]].head(20)
```

```python
# What are the extracted place tokens for unmatched segments?
unmatched = df[df["CVE_MUN"].isna()]
all_places = [p for places in unmatched["_places"] for p in places]
pd.Series(all_places).value_counts().head(40)
```

```python
# sum up road segments per municipality

df["longitud_sum"] = df.groupby("CVE_MUN")["longitud"].transform("sum")
print(df.head())
```

```python
# take the average number of lanes per municipality
df["carriles_mean"] = df.groupby("CVE_MUN")["carriles"].transform("mean")
print(df.head())
```

```python
# percentage with paid tolls
df["is_cuota"] = (df["operacion"] == "Cuota").astype(int)
df["share_cuota"] = df.groupby("CVE_MUN")["is_cuota"].transform("mean")

print(df.head())
```

```python
final_df2 = df[['CVE_MUN', 'longitud_sum', 'carriles_mean', 'share_cuota']].rename(columns={'CVE_MUN': 'CVEGEO'})

final_df2 = final_df2.groupby("CVEGEO").first().reset_index()

print(final_df2.head())

print(len(final_df2))
```
