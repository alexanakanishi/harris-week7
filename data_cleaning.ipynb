```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

df = pd.read_csv("data/Datos_viales_2013_2024.csv")
municipios = gpd.read_file("data/Mexico_mun_layer_simplified.shp")  # INEGI shapefile

# Create GeoDataFrame from your data
gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df.longitud, df.latitud),
    crs="EPSG:4326"
)

# Reproject to match shapefile CRS if needed
municipios = municipios.to_crs("EPSG:4326")

# Spatial join
result = gpd.sjoin(gdf, municipios[["NOMGEO", "geometry"]], how="left", predicate="within")
result = result.rename(columns={"NOMGEO": "municipio"})
result.drop(columns=["geometry", "index_right"])
```

```python
print(result.isna().sum())
print(len(result))
```

```python
# Are the missing-municipio rows random, or clustered?
print(result[result["municipio"].isna()]["estado"].value_counts())
print(result[result["municipio"].isna()]["ejercicio"].value_counts())
```

```python
## missing municipio rows are well balanced so we drop NA rows
result = result.dropna()
```
