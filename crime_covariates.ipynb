```python
import pandas as pd

df = pd.read_csv("data/IDM_NM_dic25.csv", encoding="latin-1")
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
month_cols = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
              'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre']

df['year_total'] = df[month_cols].sum(axis=1)
df = df.drop(columns=month_cols)

print(df.head())
print(df.columns.tolist())
```

```python
print(df['Tipo de delito'].unique())

print(df['Subtipo de delito'].unique())
```

```python
df_wide = df.groupby(
    ['Clave_Ent', 'Entidad', 'Cve. Municipio', 'Municipio',
     'Bien jurídico afectado', 'Tipo de delito', 'Subtipo de delito', 'Modalidad'],
    as_index=False
).agg(year_total=('year_total', 'sum'))

df_wide.insert(0, 'Año', 'total')

print(df_wide.head())
```
