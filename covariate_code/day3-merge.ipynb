# Merging part 1

- traffic (roads1 csv)
- ice
- mexico census infrastructure data
- schools infrastructure data


```python
import pandas as pd

df = pd.read_csv("data/covariates/ethan_final_key_df_head.csv")
traffic = pd.read_csv("data/covariates/roads1.csv")
ice = pd.read_csv("data/covariates/ice_covariates.csv")
census = pd.read_csv("data/covariates/mexico_census_infrastructure_data.csv")
schools = pd.read_csv("data/covariates/school_infrastructure_covariates.csv")
```

```python
df.head()
```

```python
traffic.head()
```

```python
traffic = traffic.drop(columns=['municipio'])
```

```python
ice.head()
```

```python
ice = ice.drop(columns = ['NOMGEO'])
```

```python
census.head()
```

```python
census = census.drop(columns = ['NOMGEO'])
```

```python
schools.head()
```

```python
schools = schools.drop(columns = ['NOMGEO'])
```

```python
print(df["CVEGEO"].nunique())
print(traffic["CVEGEO"].nunique())
print(ice["CVEGEO"].nunique())
print(census["CVEGEO"].nunique())
print(schools["CVEGEO"].nunique())
```

```python
anti_join1 = df[~df["CVEGEO"].isin(traffic["CVEGEO"])]

print(anti_join1["CVEGEO"].nunique())

anti_join2 = traffic[~traffic["CVEGEO"].isin(df["CVEGEO"])]

print(anti_join2["CVEGEO"].nunique())
print(anti_join2["CVEGEO"].unique())

print(anti_join2)
```

```python
codes_to_drop = [9998, 14998, 20999, 22998, 26998, 26999, 30998, 30999]
traffic = traffic[~traffic["CVEGEO"].isin(codes_to_drop)]
```

```python
print(df["CVEGEO"].nunique())
print(traffic["CVEGEO"].nunique())
print(ice["CVEGEO"].nunique())
print(census["CVEGEO"].nunique())
print(schools["CVEGEO"].nunique())
```

```python
merged = df.merge(traffic, on="CVEGEO", how="outer") \
           .merge(ice, on="CVEGEO", how="outer") \
           .merge(census, on="CVEGEO", how="outer") \
           .merge(schools, on="CVEGEO", how="outer")
```

```python
merged.head()
```

```python
merged.columns.tolist()
```

```python
merged.to_csv("merged_covariates.csv", index=False)
```

# Merging part 2

- federal and parastatal real estate
- roads

```python
import pandas as pd 

merged = pd.read_csv("merged_covariates.csv")
road_shape = pd.read_csv("data/covariates/municipality_road_summary.csv")
federal_palaces = pd.read_csv("data/covariates/Federal_and_Parastatal_Real_Estate_Assets_Final.csv")
```

```python
print(road_shape.head())
print(road_shape['CVEGEO'].nunique())


road_shape["CVEGEO"] = road_shape["CVEGEO"].astype(str).str.zfill(5)
road_shape = road_shape.drop(columns = ['NOMGEO'])


print("-------------")
print(road_shape.head())
```

```python
print(federal_palaces.head())
print(federal_palaces['CVEGEO'].nunique())

federal_palaces["CVEGEO"] = federal_palaces["CVEGEO"].astype(str).str.zfill(5)
federal_palaces = federal_palaces.drop(columns = ['NOMGEO'])


print('-----------')
print(federal_palaces.head())
```

```python
print(merged.head())
```

```python
merged['CVEGEO'] = merged['CVEGEO'].astype(str).str.zfill(5)


merged = merged.merge(road_shape, on="CVEGEO", how="outer") \
           .merge(federal_palaces, on="CVEGEO", how="outer")
```

```python
print(merged.head())
```

```python
merged.to_csv("merged_covariates.csv", index=False)
```

# Merging part 3

```python
aero = pd.read_csv("data/covariates/aerodrome_counts_full.csv")
drainage = pd.read_csv("data/covariates/water_drainage.csv")
capital_distance = pd.read_csv("data/covariates/municipality_capital_distances.csv")
mun_infrastructure = pd.read_csv("data/covariates/municipality_infrastructure_summary.csv")
pemex = pd.read_csv("data/covariates/municipality_pemex_distances.csv")
rail = pd.read_csv("data/covariates/municipality_rail_summary.csv")
```

```python
print("-----AERO HEAD------")
print(aero.head())

print("-----DRAINAGE HEAD------")
print(drainage.head())

print("-----CAPITAL DISTANCE HEAD------")
print(capital_distance.head())

print("-----MUNICIPALITY INFRASTRUCTURE HEAD------")
print(mun_infrastructure.head())

print("-----PEMEX HEAD------")
print(pemex.head())

print("-----RAILWAY HEAD------")
print(rail.head())
```

```python
aero = aero.drop(columns = ['NOMGEO'])
mun_infrastructure = mun_infrastructure.drop(columns = ['NOMGEO'])
rail = rail.drop(columns = ['NOMGEO'])
```

```python
print(aero["CVEGEO"].nunique())
print(drainage["CVEGEO"].nunique())
print(capital_distance["CVEGEO"].nunique())
print(mun_infrastructure["CVEGEO"].nunique())
print(pemex["CVEGEO"].nunique())
print(rail["CVEGEO"].nunique())

```

```python
aero['CVEGEO'] = aero['CVEGEO'].astype(str).str.zfill(5)
drainage['CVEGEO'] = drainage['CVEGEO'].astype(str).str.zfill(5)
capital_distance['CVEGEO'] = capital_distance['CVEGEO'].astype(str).str.zfill(5)
mun_infrastructure['CVEGEO'] = mun_infrastructure['CVEGEO'].astype(str).str.zfill(5)
pemex['CVEGEO'] = pemex['CVEGEO'].astype(str).str.zfill(5)
rail['CVEGEO'] = rail['CVEGEO'].astype(str).str.zfill(5)

```

```python
merged_final = merged.merge(aero, on="CVEGEO", how="outer") \
           .merge(drainage, on="CVEGEO", how="outer") \
           .merge(mun_infrastructure, on="CVEGEO", how="outer") \
           .merge(pemex, on="CVEGEO", how="outer") \
           .merge(rail, on="CVEGEO", how="outer")
```

```python
print(merged_final.head())
```

```python
merged_final.columns.to_list()
```

```python
merged_final.to_csv("merged_covariates.csv", index=False)
```
